```python
# So as you can see alot of words are from same root, so its better for us to replace them with same words because e.g love,loved, loving have same root word
# lov but if we keep all love loved loving model will consider them as separate word but we want them as same so our model learns that they are similar
# kinda movies 
```

```python
import nltk
```

```python
from sklearn.feature_extraction.text import CountVectorizer 
```

```python
cv =  CountVectorizer(max_features = 7000, stop_words = "english")
```

```python
vectors = cv.fit_transform(data['info']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[0:70]
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def rootWords(str):
    x = []
    for i in str.split():
        x.append(ps.stem(i))

    return " ".join(x)
```

```python
rootWords('loved')
```

```python
data['info'].apply(rootWords)
```

```python
data['info'] = data['info'].apply(rootWords)
data['info'][0]
```

```python
cv2 = CountVectorizer(max_features = 7000 ,stop_words = "english")
```

```python
vectors = cv2.fit_transform(data['info']).toarray()
```

```python
vectors
```

```python
cv2.get_feature_names_out()[0:500]
```

```python
# Now we need to calculate similarities
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vectors).shape
```

```python
cosine_similarity(vectors)
```

```python
similarities = cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarities[0])),reverse = True, key = lambda x: x[1])[1:8]
```

```python
def recommend(movie):
    movie_index = data[data['title'] == movie.lower()].index[0]
    similarity = similarities[movie_index]
    movies_list = sorted(list(enumerate(similarity)), reverse=True, key=lambda x: x[1])[1:8]
    for i in movies_list:
        print(data.iloc[i[0]].title.title())
```

```python
data.head(1)
```

```python
recommend('Avatar')
```

```python
recommend('Batman')
```

```python
recommend('titanic')
```
